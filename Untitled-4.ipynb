```python
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from scipy.signal import convolve2d

# Replace with your file path
img_path = r"C:\Users\fhabi\Desktop\Real.png"
img = Image.open(img_path).convert('L')
img = np.array(img)

# Define 3x3 filter (example: edge detection)
kernel = np.array([
   [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

# Apply convolution
feature_map = convolve2d(img, kernel, mode='same', boundary='symm')
sigmoid_map = 1 / (1 + np.exp(-feature_map))
# Plot original image and feature map
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.imshow(sigmoid_map, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(feature_map**2, cmap='gray')
plt.title("3x3 Filter Output")
plt.axis('off')

plt.show()
```
