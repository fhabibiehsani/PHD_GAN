```python
import matplotlib.pyplot as plt
import numpy as np
import os

batches = ["batch_16", "batch_8", "batch_4", "batch_2"]
batch_labels = ["Batch 16", "Batch 8", "Batch 4", "Batch 2"]
x = np.arange(len(batches))

architectures = ["U-Net", "DenseU-Net", "ResU-Net"] #, "DRU-Net"
losses = ["No SSIM", "By SSIM"]

metric_groups = [
    ["MSE", "MAE"],
    ["PSNR", "SSIM", "MS-SSIM"],
    ["FID", "LPIPS"]
]

colors = {
    "U-Net": "tab:blue",
    "DenseU-Net": "tab:orange",
    "ResU-Net": "tab:green",
    "DRU-Net": "tab:red",
}

linestyles = {
    "No SSIM": "--",
    "By SSIM": "-"
}

fig, axes = plt.subplots(3, 3, figsize=(18, 12))

for row, metrics in enumerate(metric_groups):

    for col, metric in enumerate(metrics):

        ax = axes[row, col]

        for arch in architectures:
            for loss in losses:

                y = [
                    data[batch][arch][loss][metric]
                    for batch in batches
                ]

                ax.plot(
                    x,
                    y,
                    marker='o',
                    linewidth=2,
                    color=colors[arch],
                    linestyle=linestyles[loss],
                    label=f"{arch} ({loss})"
                )

        ax.set_title(metric, fontsize=13)
        ax.set_xticks(x)
        ax.set_xticklabels(batch_labels)
        ax.set_xlabel("Batch Size")
        ax.set_ylabel(metric)
        ax.grid(True, alpha=0.3)

        # Show legend only once
        if row == 0 and col == 0:
            ax.legend(fontsize=8, ncol=2)

# Hide unused subplots
for row in range(3):
    for col in range(3):
        if col >= len(metric_groups[row]):
            axes[row, col].axis("off")

# plt.suptitle(
#     "Performance Comparison Across Batch Sizes",
#     fontsize=18
# )

plt.tight_layout(rect=[0, 0, 1, 0.96])

path = r"C:\Users\fhabi\Desktop\Results"

plt.savefig(
    os.path.join(path, "architecture_batch_comparison.png"),
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

architectures = [
    "U-Net(1_5_80)",
    "U-Net(1_5_100)",
    "U-Net(1_20_100)",
    "U-Net(1_40_50)"
]

data = {
    "NMSE": [0.0233, 0.0238, 0.0229, 0.0009],
    "NMAE": [0.0930, 0.0937, 0.0921, 0.0188],
    "PSNR": [22.7207, 22.6992, 22.6582, 36.7598],
    "SSIM": [0.6842, 0.6797, 0.6924, 0.9726],
    "MS-SSIM": [0.8538, 0.8514, 0.8557, 0.9926],
    "FID": [134.8038, 128.1669, 122.6197, 59.6118],
    "LPIPS": [0.3516, 0.3531, 0.3377, 0.0319]
}


metrics = list(data.keys())

fig, axes = plt.subplots(
    2, 4,
    figsize=(20,8)
)

axes = axes.flatten()

x = np.arange(len(architectures))


for i, metric in enumerate(metrics):

    axes[i].plot(
        x,
        data[metric],
        marker='o',
        linewidth=2,
        markersize=7
    )

    axes[i].set_title(metric, fontsize=13)

    axes[i].set_xticks(x)
    axes[i].set_xticklabels(
        architectures,
        rotation=35,
        ha="right"
    )

    axes[i].grid(True)

    # show values
    for j, v in enumerate(data[metric]):
        if not np.isnan(v):
            axes[i].text(
                j,
                v,
                f"{v:.4f}",
                ha="center",
                va="bottom",
                fontsize=8
            )


# remove empty subplot
axes[-1].axis("off")

fig.suptitle(
    "U-Net Architecture Variants Comparison (Batch Size=16)",
    fontsize=16
)

plt.tight_layout()

# plt.savefig(
#     "UNet_variants_all_metrics.png",
#     dpi=300,
#     bbox_inches="tight"
# )

plt.show()
```
